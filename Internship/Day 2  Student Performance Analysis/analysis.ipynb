```python
df = pd.read_csv("StudentsPerformance.csv")
mean = df["math score"].mean()
print("Mean:", mean)
```

```python
df=pd.read_csv("StudentsPerformance.csv")
median=df["math score"].median()
print("Median:", median)
```

```python
df=pd.read_csv("StudentsPerformance.csv")
mode=df["math score"].mode()
print("Mode:",mode[0])
```

```python
variance=df["math score"].var()
print("Variance:",variance)
```

```python
standard_deviation=df["math score"].std()
print("standard_deviation:",standard_deviation)
```

```python
q1 = df["math score"].quantile(0.25)
q2 = df["math score"].quantile(0.50)
q3 = df["math score"].quantile(0.75)

minimum = df["math score"].min()
maximum = df["math score"].max()

print("25th Percentile:", q1)
print("50th Percentile:", q2)
print("75th Percentile:", q3)
print("Minimum:", minimum)
print("Maximum:", maximum)

print("Middle 50% Range:", q1, "to", q3)
```

```python
corr1 = df["reading score"].corr(df["math score"])

print("Reading Score vs Math Score:", corr1)
```

```python
correlations = {
    "reading score": corr1,
}

highest = max(correlations, key=correlations.get)

print("Highest Correlation:", highest)

#in this code, we are performing data analysis on a dataset containing students' performance scores. We read the data from a CSV file, explore its structure and statistics, and visualize the results using histograms and scatter plots. We also calculate various statistical measures such as mean, median, mode, variance, standard deviation, percentiles, and correlations between different scores. Finally, we save the generated plots in an "images" directory.
```

```python
plt.figure(figsize=(6,4))

plt.scatter(df["reading score"], df["math score"])

plt.xlabel("Reading Score")
plt.ylabel("Math Score")
plt.title("Reading Score vs Math Score")

plt.savefig("images/scatter_plot.png")
plt.show()
```

```python
passed = (df["math score"] >= 50).mean()

print("Probability of Passing:", passed)
```

```python
score80 = (df["math score"] >= 80).mean()

print("Probability of Score >=80:", score80)
```

```python
study5 = (df["Study_Hours"] > 5).mean()

print("Probability of Study >5 Hours:", study5)
```

```python
plt.figure(figsize=(6,4))

plt.hist(df["math score"], bins=8)

plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.title("Histogram of Math Scores")

plt.savefig("images/histogram.png")
plt.show()
```

```python
df["Z_Score"] = stats.zscore(df["math score"])

print(df[["gender","math score","Z_Score"]])
```

```python
outliers_z = df[abs(df["Z_Score"]) > 2]

print(outliers_z)
```

```python
Q1 = df["math score"].quantile(0.25)
Q3 = df["math score"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["math score"] < lower) |
    (df["math score"] > upper)
]

print(outliers)
```

```python
plt.figure(figsize=(5,5))

plt.boxplot(df["math score"])

plt.title("Box Plot of Math Scores")

plt.savefig("images/boxplot.png")
plt.show()
```
